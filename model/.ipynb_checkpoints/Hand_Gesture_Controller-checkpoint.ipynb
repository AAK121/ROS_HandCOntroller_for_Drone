```python
def __write_2_csv(mode,landmark_list):
    file_path = "/home/alok/catkin_ws/src/hand_controller/model/Dataset/labelled_data.csv"
    with open(file_path,'a',newline="") as f:
        writer = csv.writer(f)
        writer.writerow([mode, *landmark_list])
    return

def __write_2_csv_processed(mode,landmark_list):
    file_path = "/home/alok/catkin_ws/src/hand_controller/model/Dataset/labelled_data_processed.csv"
    with open(file_path,'a',newline="") as f:
        writer = csv.writer(f)
        writer.writerow([mode, *landmark_list])
    return
```

```python

def get_landmarks(image,hand_landmarks):
    landmark_list = []
    img_height,img_width = image.shape[0],image.shape[1]
    for landmarks in hand_landmarks.landmark:
        x = min(int(landmarks.x*img_width),img_width-1)
        y = min(int(landmarks.y*img_height),img_height-1)
        landmark_list.append([x,y])
    return landmark_list

def process_landmarks(landmarks):
    relative_x,relative_y = 0,0
    temp_landmarks = copy.deepcopy(landmarks)
    final_landmarks = []
    for index, point in enumerate(temp_landmarks):
        if index == 0:
            relative_x,relative_y = point[0],point[1]
        temp_landmarks[index][0] = point[0] - relative_x
        temp_landmarks[index][1] = point[1] - relative_y
        final_landmarks.append(temp_landmarks[index][0])
        final_landmarks.append(temp_landmarks[index][1])
    max_value = max(list(map(abs,final_landmarks)))

    def normalize(n):
        return n/max_value

    final_landmarks = list(map(normalize,final_landmarks))

    return final_landmarks
    
        
```

```python
# Initialize MediaPipe Hands and Drawing modules
mp_hands = mp.solutions.hands
mp_drawing = mp.solutions.drawing_utils
mp_drawing_styles = mp.solutions.drawing_styles

# Open the webcam
cap = cv.VideoCapture(0)
if not cap.isOpened():
    print("Error: Could not open the webcam.")
    exit()

# Initialize MediaPipe Hands with default settings
with mp_hands.Hands(
    model_complexity=1,
    min_detection_confidence=0.5,
    min_tracking_confidence=0.5
) as hands:
    # mode = 0
    # while True:
    #     number = 0
    number = 0
    mode = 0
    while True:
        ret, frame = cap.read()
        if not ret:
            print("Error: Could not capture frame.")
            break

        # Flip the frame horizontally for a mirror-like effect
        frame = cv.flip(frame, 1)

        # Convert the frame to RGB (required by MediaPipe)
        rgb_frame = cv.cvtColor(frame, cv.COLOR_BGR2RGB)

        # Process the frame to detect hands
        results = hands.process(rgb_frame)
        landmarks_lists = []
        # Draw hand landmarks if detected
        if results.multi_hand_landmarks:
            for hand_landmarks in results.multi_hand_landmarks:
                mp_drawing.draw_landmarks(
                    frame,
                    hand_landmarks,
                    mp_hands.HAND_CONNECTIONS,
                    mp_drawing_styles.get_default_hand_landmarks_style(),
                    mp_drawing_styles.get_default_hand_connections_style()
                )
                landmarks_list = get_landmarks(frame,hand_landmarks)
                processed_lanamrks = process_landmarks(landmarks_list)
                landmarks
                

        # Display the frame
        cv.imshow("Hand Landmarks", frame)

        key  = cv.waitKey(1) & 0xFF
        # Break the loop if 'q' is pressed
        if key == ord('q'):
            print("Exiting...")
            break
        elif key == ord('c'):
                    print(f"Data No.: {number}")
                    __write_2_csv_processed(mode,processed_lanamrks)
                    number +=1
            
        # mode+=1        
        # if mode == 10:
        #     print("All modes Done")
        #     break

# Release resources
cap.release()
cv.destroyAllWindows()
```

```python
import cv2 as cv 
import mediapipe as mp
import copy 
import csv

cap = cv.VideoCapture(0)

# Initialize MediaPipe Hands and Drawing utilities
Hands = mp.solutions.hands
hands = Hands.Hands(
    static_image_mode=False,
    max_num_hands=1,
    min_detection_confidence=0.7,
    min_tracking_confidence=0.7,
)
mp_draw = mp.solutions.drawing_utils

if not cap.isOpened():
    print("Error opening camera")
else:
    for mode in range(1,10):
        number = 0
        while True:
            ret, frame = cap.read()
            if not ret:
                print("Error capturing")
                break
            else:
                # Flip the frame for mirror-like effect
                image = cv.flip(frame, 1)
                copy_image = copy.deepcopy(image)
    
                # Convert to RGB for MediaPipe processing
                rgb_image = cv.cvtColor(image, cv.COLOR_BGR2RGB)
                results = hands.process(rgb_image)
                processed_landmarks_final = []
                # Draw hand landmarks if detected
                if results.multi_hand_landmarks:
                    for landmarks in results.multi_hand_landmarks:
                        mp_draw.draw_landmarks(
                            copy_image,  # Draw on the BGR frame
                            landmarks,
                            Hands.HAND_CONNECTIONS
                        )
                        landmarks = get_landmarks(frame,landmarks)
                        processed_landmarks = process_landmarks(landmarks)
                        processed_landmarks_final.extend(processed_landmarks)
                else:
                    print("No hands detected!")
    
                # Show the result
                cv.imshow("Camera Frame", copy_image)
    
                # Exit loop if 'q' is pressed
                key = cv.waitKey(1) & 0xFF
                if key == ord("q"):
                    print("Exiting")
                    break
                elif key == ord('c'):
                    print(f"Data No.: {number}")
                    __write_2_csv_processed(0,processed_landmarks_final)
                    number+=1
                

# Release resources
cap.release()
cv.destroyAllWindows()
```
